```python
#1BF24CS181-Muddam Rishitha Reddy
#GOAL BASED

def vacuum_agent(roomA, roomB, location):

    while roomA == "Dirty" or roomB == "Dirty":

        if location == "A":
            if roomA == "Dirty":
                print("Room A is dirty -> Suck")
                roomA = "Clean"
            else:
                print("Room A is clean -> Move Right")
                location = "B"

        else: # location == B
            if roomB == "Dirty":
                print("Room B is dirty -> Suck")
                roomB = "Clean"
            else:
                print("Room B is clean -> Move Left")
                location = "A"

    print("Goal achieved: Both rooms are clean")


# Take initial state from user
roomA = input("Enter status of Room A (Clean/Dirty): ")
roomB = input("Enter status of Room B (Clean/Dirty): ")
location = input("Enter vacuum location (A/B): ")

vacuum_agent(roomA, roomB, location)
```
