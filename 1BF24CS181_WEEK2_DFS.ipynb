```python
def vacuum_agent(roomA, roomB, location):

    goal = "Both rooms clean"

    while roomA == "Dirty" or roomB == "Dirty":

        if location == "A":
            if roomA == "Dirty":
                print("Room A is dirty → Suck")
                roomA = "Clean"
            else:
                print("Room A is clean → Move Right")
                location = "B"

        elif location == "B":
            if roomB == "Dirty":
                print("Room B is dirty → Suck")
                roomB = "Clean"
            else:
                print("Room B is clean → Move Left")
                location = "A"

    print("Goal achieved:", goal)


# Initial state
vacuum_agent("Dirty", "Dirty", "A")
```
