```python
#1BF24CS181-Muddam Rishitha Reddy

GOAL_STATE = (1, 2, 3, 4, 5, 6, 7, 8, 0)
INITIAL_STATE = (1, 2, 3, 0, 4, 6, 7, 5, 8)


def get_neighbors(state):
    neighbors = []

    idx = state.index(0)
    row, col = idx // 3, idx % 3

    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nr, nc = row + dr, col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_idx = nr * 3 + nc

            lst = list(state)
            lst[idx], lst[new_idx] = lst[new_idx], lst[idx]

            neighbors.append(tuple(lst))

    return neighbors


def print_grid(state):
    for r in range(3):
        row = []
        for x in state[r * 3:(r + 1) * 3]:
            row.append(str(x) if x != 0 else "_")
        print(" ".join(row))
    print()


def dfs(initial_state, goal_state):
    stack = [(initial_state, [])]
    visited = set()

    while stack:

        state, path = stack.pop()

        if state in visited:
            continue

        visited.add(state)

        print("\nPARENT:")
        print_grid(state)

        if state == goal_state:
            print("GOAL REACHED!")
            print("\nSolution Path:")

            for i, s in enumerate(path + [state]):
                print("Step", i)
                print_grid(s)

            return

        neighbors = get_neighbors(state)

        # DFS: add neighbors to stack
        valid_neighbors = [
            n for n in neighbors
            if n not in visited
        ]

        print("POSSIBILITIES:")

        for n in valid_neighbors:
            print_grid(n)

        # Reverse so the first neighbor is explored first
        for n in reversed(valid_neighbors):
            stack.append((n, path + [state]))

    print("Goal not found!")


if __name__ == "__main__":
    dfs(INITIAL_STATE, GOAL_STATE)
```
