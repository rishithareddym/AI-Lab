```python

#1BF24CS181
#Muddam Rishitha Reddy
import heapq

# Manhattan Distance
def heuristic(state, goal):
    h = 0

    for i in range(9):
        if state[i] == 0:
            continue

        r1 = i // 3
        c1 = i % 3

        j = goal.index(state[i])
        r2 = j // 3
        c2 = j % 3

        h += abs(r1 - r2) + abs(c1 - c2)

    return h


# Generate children
def get_children(state):
    children = []

    blank = state.index(0)
    row = blank // 3
    col = blank % 3

    # Left
    if col > 0:
        new_state = list(state)
        new_state[blank], new_state[blank - 1] = \
            new_state[blank - 1], new_state[blank]
        children.append((tuple(new_state), "L"))

    # Right
    if col < 2:
        new_state = list(state)
        new_state[blank], new_state[blank + 1] = \
            new_state[blank + 1], new_state[blank]
        children.append((tuple(new_state), "R"))

    # Up
    if row > 0:
        new_state = list(state)
        new_state[blank], new_state[blank - 3] = \
            new_state[blank - 3], new_state[blank]
        children.append((tuple(new_state), "U"))

    # Down
    if row < 2:
        new_state = list(state)
        new_state[blank], new_state[blank + 3] = \
            new_state[blank + 3], new_state[blank]
        children.append((tuple(new_state), "D"))

    return children


# A* Search
def a_star(start, goal):

    # Priority Queue: f, g, state, path
    frontier = []

    g = 0
    h = heuristic(start, goal)
    f = g + h

    heapq.heappush(frontier, (f, g, start, []))

    visited = {}

    while frontier:

        # Get node with minimum f = g + h
        f, g, state, path = heapq.heappop(frontier)

        # Goal test
        if state == goal:
            return path, g

        if state in visited and visited[state] <= g:
            continue

        visited[state] = g

        # Generate children
        for child, move in get_children(state):

            new_g = g + 1
            new_h = heuristic(child, goal)
            new_f = new_g + new_h

            new_path = path + [move]

            heapq.heappush(
                frontier,
                (new_f, new_g, child, new_path)
            )

    return None, -1


# ---------------- MAIN ----------------

print("Enter initial state:")
start = tuple(map(int, input().split()))

print("Enter goal state:")
goal = tuple(map(int, input().split()))

path, cost = a_star(start, goal)

if path is not None:
    print("\nSolution found!")
    print("Moves:", path)
    print("Number of moves:", cost)
else:
    print("No solution")
```
