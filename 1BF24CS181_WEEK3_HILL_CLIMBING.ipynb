```python
import random

def get_conflicts(state):
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j]:
                conflicts += 1
            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_best_neighbor(state):
    n = len(state)
    current_conflicts = get_conflicts(state)
    best_state = list(state)
    best_conflicts = current_conflicts

    print("Possibilities and h-values:")
    for col in range(n):
        original_row = state[col]
        for row in range(n):
            if row == original_row:
                continue
            neighbor = list(state)
            neighbor[col] = row
            neighbor_conflicts = get_conflicts(neighbor)
            print(f"Move queen in col {col} to row {row}: state={neighbor}, h={neighbor_conflicts}")
            if neighbor_conflicts < best_conflicts:
                best_conflicts = neighbor_conflicts
                best_state = neighbor
    return best_state, best_conflicts

def hill_climbing_4_queens_user_input(initial_state):
    n = 4
    state = list(initial_state)
    current_conflicts = get_conflicts(state)
    restarts = 0

    while True:
        while True:
            print(f"Current state={state}, h={current_conflicts}")
            neighbor, neighbor_conflicts = get_best_neighbor(state)
            if neighbor_conflicts >= current_conflicts:
                print(f"Stuck in local minimum at state={state} with h={current_conflicts}")
                break
            state = neighbor
            current_conflicts = neighbor_conflicts

        if current_conflicts == 0:
            return state, restarts

        state = [random.randint(0, n - 1) for _ in range(n)]
        current_conflicts = get_conflicts(state)
        restarts += 1
        print(f"Restarting with random state={state}, h={current_conflicts}")

def print_board(state):
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += "[ Q ] "
            else:
                line += "[ . ] "
        print(line)
    print()

if __name__ == "__main__":
    while True:
        raw_input = input("Your initial state (or type 'stop' to quit): ").strip().lower()
        if raw_input == 'stop':
            break

        try:
            user_state = [int(x) for x in raw_input.split()]
            if len(user_state) != 4 or any(row < 0 or row > 3 for row in user_state):
                print("Invalid input")
                continue
        except ValueError:
            print("Invalid input")
            continue

        print(f"Initial positions: {user_state}")
        print_board(user_state)

        solution, total_restarts = hill_climbing_4_queens_user_input(user_state)

        print(f"Solution found with {total_restarts} restarts")
        print(f"Final positions: {solution}")
        print_board(solution)
```
