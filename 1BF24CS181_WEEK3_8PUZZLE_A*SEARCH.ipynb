```python

#1BF24CS181
#Muddam Rishitha Reddy
import heapq

class PuzzleState:
    def __init__(self, board, parent=None, move="", depth=0, cost=0, h_val=0):
        self.board = board
        self.parent = parent
        self.move = move
        self.depth = depth
        self.h_val = h_val
        self.cost = cost

    def __lt__(self, other):
        return self.cost < other.cost

def get_misplaced_tiles(board, goal):
    misplaced = 0
    for i in range(len(board)):
        if board[i] != 0 and board[i] != goal[i]:
            misplaced += 1
    return misplaced

def get_neighbors(state):
    neighbors = []
    zero_idx = state.board.index(0)
    row, col = divmod(zero_idx, 3)

    moves = {
        'Up': (-1, 0, -3),
        'Down': (1, 0, 3),
        'Left': (0, -1, -1),
        'Right': (0, 1, 1)
    }

    for move_name, (dr, dc, offset) in moves.items():
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_zero_idx = zero_idx + offset
            new_board = list(state.board)
            new_board[zero_idx], new_board[new_zero_idx] = new_board[new_zero_idx], new_board[zero_idx]
            neighbors.append((tuple(new_board), move_name))

    return neighbors

def solve_8_puzzle(start_state, goal_state):
    open_list = []
    visited = set()

    h_start = get_misplaced_tiles(start_state, goal_state)
    root = PuzzleState(board=start_state, parent=None, move="", depth=0, cost=h_start, h_val=h_start)
    heapq.heappush(open_list, root)

    while open_list:
        current_state = heapq.heappop(open_list)

        if current_state.board == goal_state:
            return reconstruct_path(current_state)

        visited.add(current_state.board)

        for next_board, move_name in get_neighbors(current_state):
            if next_board in visited:
                continue

            new_depth = current_state.depth + 1
            h_cost = get_misplaced_tiles(next_board, goal_state)
            f_cost = new_depth + h_cost

            child_node = PuzzleState(
                board=next_board,
                parent=current_state,
                move=move_name,
                depth=new_depth,
                h_val=h_cost,
                cost=f_cost
            )
            heapq.heappush(open_list, child_node)

    return None

def reconstruct_path(state):
    path = []
    curr = state
    while curr.parent is not None:
        path.append((curr.move, curr.board, curr.depth, curr.h_val))
        curr = curr.parent
    path.reverse()
    return path

def print_board(board):
    for i in range(0, 9, 3):
        print(f"[ {board[i]} {board[i+1]} {board[i+2]} ]")
    print()

def get_user_board_input(prompt_message):
    while True:
        try:
            print(prompt_message)
            user_input = input("Enter 9 numbers separated by spaces (0 for empty): ").strip()

            board = tuple(int(x) for x in user_input.split())

            if len(board) != 9:
                print("Error: You must enter exactly 9 numbers.\n")
                continue
            if sorted(board) != list(range(9)):
                print("Error: Board must contain all digits from 0 to 8 without duplicates.\n")
                continue

            return board
        except ValueError:
            print("Error: Invalid characters detected. Please use integers only.\n")

if __name__ == "__main__":
    print("8-PUZZLE A* SOLVER (MISPLACED TILES)\n")

    initial = get_user_board_input("Define Initial State")
    print("\nInitial configuration saved:")
    print_board(initial)

    goal = get_user_board_input("Define Goal State")
    print("\nGoal configuration saved:")
    print_board(goal)

    print("Searching for an optimal path...\n")
    solution = solve_8_puzzle(initial, goal)

    if solution:
        print(f"Goal reached in {len(solution)} steps.\n")
        for step, (move, board, g, h) in enumerate(solution, 1):
            print(f"Step {step} - Move blank space: {move} | g(n) = {g}, h(n) = {h}")
            print_board(board)
    else:
        print("This puzzle configuration is mathematically unsolvable from the given starting point.")
```
