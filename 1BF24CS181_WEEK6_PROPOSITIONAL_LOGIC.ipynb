```python
#1BF24CS181--Muddam Rishitha Reddy

import re

def PL_TRUE(sentence, model):
    return eval(sentence, {"__builtins__": None}, model)


def TT_ENTAILS(KB, alpha):
    symbols = set()

    for sentence in KB + [alpha]:
        symbols.update(re.findall(r'\b[A-Z]\b', sentence))

    return TT_CHECK_ALL(KB, alpha, list(symbols), {})


def TT_CHECK_ALL(KB, alpha, symbols, model):

    if len(symbols) == 0:
        if all(PL_TRUE(sentence, model) for sentence in KB):
            return PL_TRUE(alpha, model)
        else:
            return True

    P = symbols[0]
    rest = symbols[1:]

    return (
        TT_CHECK_ALL(KB, alpha, rest, {**model, P: True})
        and
        TT_CHECK_ALL(KB, alpha, rest, {**model, P: False})
    )


# Taking input from user
n = int(input("Enter number of sentences in KB: "))

KB = []
for i in range(n):
    sentence = input(f"Enter KB sentence {i+1}: ")
    KB.append(sentence)

alpha = input("Enter query (alpha): ")

if TT_ENTAILS(KB, alpha):
    print("KB entails alpha")
else:
    print("KB does not entail alpha")
```
