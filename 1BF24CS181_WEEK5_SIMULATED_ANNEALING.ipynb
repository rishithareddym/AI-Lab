```python
#1BF24CS181-Muddam Rishitha Reddy

import random
import math


# Calculate number of conflicts
def calculate_cost(board):
    cost = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):

            # Same row
            if board[i] == board[j]:
                cost += 1

            # Same diagonal
            elif abs(board[i] - board[j]) == abs(i - j):
                cost += 1

    return cost


# Generate a neighboring solution
def get_neighbor(board):
    new_board = board.copy()

    # Pick a random column
    col = random.randint(0, 7)

    # Pick a different random row
    new_row = random.randint(0, 7)

    while new_row == new_board[col]:
        new_row = random.randint(0, 7)

    new_board[col] = new_row

    return new_board


# Simulated Annealing
def simulated_annealing():

    # Initial random board
    board = [random.randint(0, 7) for _ in range(8)]

    current_cost = calculate_cost(board)

    # Initial temperature
    T = 10.0

    # Cooling factor
    alpha = 0.90

    # Continue until temperature becomes very small
    while T > 0.001:

        # If solution is found
        if current_cost == 0:
            return board, current_cost

        # Generate neighbor
        new_board = get_neighbor(board)

        # Calculate new cost
        new_cost = calculate_cost(new_board)

        # Difference in cost
        delta_E = new_cost - current_cost

        # Better solution
        if delta_E < 0:
            board = new_board
            current_cost = new_cost

        # Worse solution
        else:
            probability = math.exp(-delta_E / T)

            random_value = random.random()

            if random_value < probability:
                board = new_board
                current_cost = new_cost

        # Cool down
        T = T * alpha

    return board, current_cost


# Run the algorithm
solution, cost = simulated_annealing()

print("Final Board:", solution)
print("Final Cost:", cost)
```
